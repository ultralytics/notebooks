# Ultralytics Construction-PPE Detection with YOLO26

Train YOLO26 on the [Ultralytics Construction-PPE dataset](https://docs.ultralytics.com/datasets/detect/construction-ppe): 1,000 images with 12 classes for people, forklifts, protective equipment, and visible missing PPE.

The dataset was created and open-sourced in November 2026 by the Ultralytics team. It combines human-collected and generated images, with expert human annotation and AI auto-annotation completed in November 2026 by the Ultralytics team. Images and annotations are released under [CC BY 4.0](https://creativecommons.org/licenses/by/4.0/).

## Dataset Structure

The November 2026 release has a fixed 80/20 split:

- **Training**: 800 images with YOLO bounding-box annotations.
- **Validation**: 200 images with YOLO bounding-box annotations.

There is no separate test split. Images include landscape, portrait, and square compositions at varied resolutions. Use independent deployment images for final evaluation.

[Browse the dataset on Ultralytics Platform](https://platform.ultralytics.com/ultralytics/datasets/construction-ppe).

## Applications

Construction-PPE detection using computer vision improves workplace safety and compliance. By analyzing site footage, AI systems check if workers wear helmets, vests, or gloves, enabling real-time monitoring and instant alerts. These systems can also provide safety analytics, support drone or robot inspections, and serve as valuable tools for research and education in workplace safety.

## Dataset YAML File

The [Construction-PPE YAML](https://github.com/ultralytics/ultralytics/blob/main/ultralytics/cfg/datasets/construction-ppe.yaml) defines the image paths and 12-class order. The dataset archive extracts into `construction-ppe`.

## Train

Train YOLO26 on [Detect](https://docs.ultralytics.com/tasks/detect), [Segment](https://docs.ultralytics.com/tasks/segment), [Semantic](https://docs.ultralytics.com/tasks/semantic), [Depth](https://docs.ultralytics.com/tasks/depth), [Classify](https://docs.ultralytics.com/tasks/classify), [Pose](https://docs.ultralytics.com/tasks/pose) and [OBB](https://docs.ultralytics.com/tasks/obb) datasets. See [YOLO26 Train Docs](https://docs.ultralytics.com/modes/train) for more information.

````
from ultralytics import YOLO

# Load a model
model = YOLO("yolo26n.pt")  # load a pretrained model (recommended for training)

# Train the model
results = model.train(
    data="https://github.com/ultralytics/assets/releases/download/v0.0.0/construction-ppe.zip",
    epochs=5,
    imgsz=640,
)
````

## Predict

Run the trained model on a validation image. The class IDs for helmet, vest, goggles, and person are `0`, `2`, `4`, and `5`. See the [prediction guide](https://docs.ultralytics.com/modes/predict) for more options.

````
from pathlib import Path

from ultralytics import YOLO

modelp = YOLO(f"{model.trainer.save_dir}/weights/best.pt")
image = next((Path(model.trainer.data["path"]) / "images" / "val").glob("*.jpg"))
prediction_results = modelp.predict(image, save=True, classes=[0, 2, 4, 5])
prediction_results[0].show()
````
